Plotly is a Python library that allows you to create interactive visualizations. 
For this demonstration, we'll be using Plotly Express, a simple interface to the capabilities of Plotly.

We will first read in our data using pandas, then we will plot data using plotly. 
Plotly express is imported as `import plotly.express as px`

```python
import pandas as pd

import plotly.express as px
```

```python
# Load data/PubChemElements_all.csv into a DataFrame named df using pandas.
# Display the first five rows to check that the data loaded correctly.
```

To plot something using plotly, we can use `px.line` or `px.scatter` for line and scatter plots respectively.

```python
fig = px.line(df, x="AtomicNumber", y="MeltingPoint")
fig.update_traces(name="Melting Point", showlegend=True)
fig.add_scatter(x=df["AtomicNumber"], y=df["BoilingPoint"], name="Boiling Point")
fig.show()
```

Once you have a variable representing a figure, you can continue to update it in different cells.
This figure includes both melting and boiling points, so we will label the y axis "Temperature (K)".


```python
fig.update_yaxes(title="Temperature (K)")
```

Plotly allows you to set the color of your markers based on the value in another column. In the cell below, we color the points based on their block.

```python
fig = px.scatter(df, x="AtomicNumber", y="IonizationEnergy", color="GroupBlock")
fig.show()
```

We can also color elements by their period (row) in the periodic table.
The example below uses [pandas `cut`](https://pandas.pydata.org/docs/reference/api/pandas.cut.html) to group atomic numbers into intervals.
Each interval ends at the last atomic number in a period: 2, 10, 18, 36, 54, 86, and 118.
For example, atomic numbers 3 through 10 receive the label "2".
We save these labels in a column called `Period`.

This is an optional data-preparation example; you are not expected to write it from memory.
The following plot uses `color="Period"` to color by those labels and `hover_data=["Name"]` to show element names when you hover over points.


```python
df["Period"] = pd.cut(
    df["AtomicNumber"],
    bins=[0, 2, 10, 18, 36, 54, 86, 118],
    labels=["1", "2", "3", "4", "5", "6", "7"],
)
```

```python
fig = px.scatter(df, x="AtomicNumber", y="IonizationEnergy", color="Period", hover_data=["Name"])
fig.show()
```

We can also do thing like visualize boiling points by standard states using a box plot.

```python
fig = px.box(df, x="StandardState", y="BoilingPoint")
fig.show()
```
