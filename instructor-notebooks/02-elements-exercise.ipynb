```python
import pandas as pd
df = pd.read_csv(ROOT / "data/PubChemElements_all.csv")
groups = df.groupby("GroupBlock")
groups.describe()
```

```python
electronegativity_stats = groups["Electronegativity"].describe()
electronegativity_stats
```

Halogens have the highest group mean (2.992); alkali metals have the lowest (0.840). Nonmetals generally have higher electronegativity than metals, consistent with their greater attraction for bonding electrons. Group means hide variation within each category. Missing values are excluded: the noble-gas mean of 2.8 uses only two values, so it should not be generalized to every noble gas.
