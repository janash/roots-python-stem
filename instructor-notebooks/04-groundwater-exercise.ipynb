```python
import pandas as pd
import statsmodels.formula.api as smf
df = pd.read_csv(ROOT / "data/ground_water.csv")
df.head()
```

```python
regression = smf.ols("pH ~ Bicarb", data=df).fit()
regression.summary()
```

```python
print(regression.params)
print("R-squared:", regression.rsquared)
print("Observations used:", int(regression.nobs))
```

The fitted equation is pH ≈ 8.0976 − 0.003052 × Bicarb (ppm). A 100 ppm increase in bicarbonate corresponds to a fitted decrease of about 0.305 pH units. R² ≈ 0.115 means the model explains about 11.5% of the observed pH variation. This association alone does not establish causation, and bicarbonate alone leaves most variation unexplained. Retain an intercept: pH need not be zero when bicarbonate is zero.
