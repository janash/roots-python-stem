```python
deltaH = 1.52  # kJ/mol
deltaS = 0.0051  # kJ/(mol K)
temperatures = [100, 150, 200, 250, 300, 350, 400]
deltaG_list = []
spontaneous_temp_list = []
for temperature in temperatures:
    deltaG = deltaH - temperature * deltaS
    deltaG_list.append(deltaG)
    if deltaG < 0:
        spontaneous_temp_list.append(temperature)
print("Delta G (kJ/mol):", deltaG_list)
print("Spontaneous temperatures (K):", spontaneous_temp_list)
```

ΔG values are 1.01, 0.755, 0.500, 0.245, −0.010, −0.265, and −0.520 kJ/mol. The reaction is spontaneous at 300, 350, and 400 K under this model. Since ΔS is positive, increasing T lowers ΔG. The crossover is ΔH/ΔS ≈ 298.04 K; ΔG = 0 is the equilibrium boundary.
