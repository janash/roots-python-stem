```python
import pandas as pd
import plotly.express as px
df = pd.read_csv(ROOT / "data/PubChemElements_all.csv")
```

```python
fig = px.scatter(df, x="AtomicNumber", y="AtomicRadius", color="GroupBlock", symbol="GroupBlock",
                 hover_data=["Name"], title="Atomic radius across the periodic table",
                 labels={"AtomicNumber": "Atomic number", "AtomicRadius": "Atomic radius (pm)", "GroupBlock": "Element group"})
fig.update_traces(marker=dict(size=9, line=dict(width=1, color="#222222")))
fig.show()
```

```python
fig = px.box(df, x="StandardState", y="BoilingPoint", title="Boiling points by standard state",
             labels={"StandardState": "Standard state", "BoilingPoint": "Boiling point (K)"})
fig.show()
```

Atomic radius shows repeating variation across atomic number, generally decreasing across a period and increasing down a group. A scatter plot preserves the individual element measurements. The box plot compares distributions: median boiling points are 87.3 K for gases, 480.915 K for liquids, and 3107.5 K for solids in this dataset. Missing measurements are omitted, including categories containing only missing values. Use the tables below for a text alternative to the plots.
